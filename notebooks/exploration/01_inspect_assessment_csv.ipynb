# Inspect Assessment CSV

## Background

Edmonton publishes annual property assessment data for all ~440k parcels in the city. Each row is one property with its assessed value, neighbourhood, tax class, and assessment classification. The goal of this project is to compute **assessed value per acre** by neighbourhood — a proxy for how fiscally productive each part of the city is.

This notebook is a one-time inspection to confirm column names, types, and quirks before writing the pipeline module (`src/load_assessment.py`).

**Key columns we care about:**
- `Neighbourhood` — join key to boundary polygons (needs normalization)
- `Assessed Value` — our main metric
- `Assessment Class 1` — used to proxy tax-exempt status (no explicit exempt flag in this dataset)

**What we found:**
- 439,769 rows total; 46 zero-value rows dropped; 3 tax-exempt rows flagged (`NONRES MUNICIPAL/RES EDUCATION`)
- `Suite` column has mixed types — always load with `low_memory=False`
- No explicit tax-exempt boolean; best proxy is `Assessment Class 1 == 'NONRES MUNICIPAL/RES EDUCATION'`
